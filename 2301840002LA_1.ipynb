Articles Preprocessing

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')

def preprocess_article(article):
  soup=BeautifulSoup(article,'html.parser')
  paragraphs=soup.find_all('p')
  article=' '.join([p.get_text() for p in paragraphs])

  article=article.lower() #Lowercase
  article=re.sub(r'<.*?>','',article) #Remove HTML Tags
  article=re.sub(r'[^a-zA-Z0-9\s]','',article) #Remove punctuation and non-alphabetic chars

  words=word_tokenize(article)#Tokenization of words

  stop_words=set(stopwords.words('english'))
  tokens=[word for word in words if word not in stop_words and len(word)>1] #Remove stop words

  return tokens # Return the processed tokens
```

```python
all_articles_tokens=[]
for article in all_articles:
  tokens=preprocess_article(article)
  all_articles_tokens.append(tokens)
```
